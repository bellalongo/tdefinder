##### SDSS Data

```python
import requests

# Target coordinates in decimal degrees
ra = 5.986917
dec = -14.423167

# API endpoint (DR17 is a standard photometric release; DR18 is also available)
url = "https://sdss.org"

params = {
    "ra": ra,
    "dec": dec,
    "scale": 0.396,  # Arcseconds per pixel (0.396 is SDSS native resolution)
    "width": 512,    # Image width in pixels
    "height": 512,   # Image height in pixels
    "opt": ""        # Optional overlays: 'G' for grid, 'L' for labels, 'P' for photo objects
}

response = requests.get(url, params=params)

if response.status_code == 200:
    with open("sdss_cutout.jpg", "wb") as f:
        f.write(response.content)
    print("Image saved successfully.")
else:
    print(f"Error {response.status_code}: Could not retrieve image.")
```

```python
from astropy import coordinates as coords
import astropy.units as u
from astroquery.sdss import SDSS

pos = coords.SkyCoord(ra=ra * u.deg, dec=dec * u.deg, frame='icrs')

# Only fetch the minimal fields needed to locate the image
xid = SDSS.query_region(pos, radius=5 * u.arcsec, photoobj_fields=['run', 'rerun', 'camcol', 'field'])

if xid is not None:
    # Slice xid[:1] so it ONLY downloads the first matching frame instead of all of them
    images = SDSS.get_images(matches=xid[:1], band='r')
    images.writeto("sdss_r_band.fits", overwrite=True)
    print("FITS file saved.")
```
